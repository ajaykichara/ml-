```python
# ── Normal Distribution ───────────────────────────────────────────────────────
# • Mean, Median, Mode are approximately equal
# • Bell-shaped and symmetric about the center
#   Symmetric = Left side of bell == Right side of bell (mirror image)
#
#              ●
#            ●   ●
#          ●       ●
#        ●           ●
#       ●             ●
#   ────────────────────────
#   Left side  |  Right side
#    (equal)  Mean  (equal)
#
# • Mean sits at center → divides bell into 2 equal halves
# • Equal distance on both sides:
#   Left side  →  -20, -10  from mean
#   Right side →  +10, +20  from mean
# • That is why Mean = Median = Mode  (all sit at center) 
# • 68% data within 1σ, 95% within 2σ, 99.7% within 3σ  (Empirical Rule)

# ── Why Convert to Standard Normal? ──────────────────────────────────────────
# • Different numeric columns have different scales
#   age          →  15, 25, 60, 75      (range 0-100)
#   fever        →  98, 101, 104        (range 98-106)
#   num_symptoms →  1, 2, 3, 4, 5      (range 1-10)
# • Model thinks bigger number = more important  (very impotant see  here )
# • So we convert all to same scale              

# ── What is x ? ──────────────────────────────────────────────────────────────
# • x = each individual data point in the column

# ── Z-Score Formula (Standardization) ────────────────────────────────────────
# • Need : Mean (μ)  and  Standard Deviation (σ)
# • Formula : z = (x - μ) / σ
#   Step 1 → (x - μ)  :  subtracts mean        →  shifts center to 0
#   Step 2 → (÷  σ )  :  divides by std         →  scales spread to 1

# ── Standard Normal Distribution (Output of Z-score) ─────────────────────────
# • Mean (μ)           ≈ 0
# • Standard Deviation ≈ 1
# • All features now on same scale  

# ── Why Scale Only Numeric Data? ─────────────────────────────────────────────
# • Categorical data  →  already encoded to 0 and 1  →  No Scaling needed 
# • Numeric data      →  different ranges             →  Scaling needed    

# ── StandardScaler (sklearn) ─────────────────────────────────────────────────
# • sc.fit(x_train)        →  learns μ and σ from data
# • sc.transform(x_train)  →  applies z = (x - μ) / σ
# • Output                 →  Standard Normal Distribution (μ ≈ 0, σ ≈ 1) not  exact  its  near to  it  
```

```python
import numpy as np
import pandas as pd

df= pd.read_csv("covid_toy.csv")
df.head(2)
```

```python
df.isnull().sum()
```

```python
df = df.dropna()
df.isnull().sum()
```

```python
df = df.drop(columns=['city','cough'])
df.head(2)
```

```python
df['gender'] = df['gender'].map({"Male":0 , "Female":1})
df['has_covid'] = df['has_covid'].map({"No":0 , "Yes":1})
# df.head(2)
```

```python
x = df.drop(columns = ['has_covid'])
y = df['has_covid']

from sklearn.model_selection import train_test_split

x_train , x_test , y_train , y_test = train_test_split(x,y,test_size=0.2,random_state=42)

np.round(x_train.describe(),2)
```

```python
# ── Step 5 : Feature Scaling (Only Numeric Data) ──────────────────────────────
# Normal Distribution  →  z = (x - μ) / σ  →  Standard Normal (μ=0, σ=1)
from sklearn.preprocessing import StandardScaler

sc = StandardScaler()

x_train_sc = sc.fit_transform(x_train)   ### fit means learn the parameters and transform means apply on that data 
# fit = learn μ,σ  |  transform = apply z formula
```
