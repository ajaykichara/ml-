```python
# The FunctionTransformer is a tool in scikit-learn, a popular Python library for machine learning,
# that allows you to apply a specified function to the input data.

# The FunctionTransformer can be useful for performing custom transformations of input data in a machine learning pipeline.
```

```python
# The FunctionTransformer takes as input a single function that will be applied to each sample in the data. 
# This function can be any Python function that takes a single argument, such as a lambda function or a user-defined function. 
# The function should return the transformed sample.
# The FunctionTransformer takes as input a single function that will be applied to each sample in the data.
# This function can be any Python function that takes a single argument, such as a lambda function or a user-defined function.
# The function should return the transformed sample.
```

```python
from sklearn.preprocessing  import FunctionTransformer
import numpy as np 

#  create data set  
x = np.array([[1,2],[3,4]])

#define  the  transformation  function  
log_transform  =  FunctionTransformer(np.log1p)

# apply  the transformation  to  the  data set  
x_transformed   =  log_transform.transform(x)

# view the  transformed  data
print(x_transformed)
```

```python
# second  example  
#create data set 
z = np.array([[1,2],[3,4]])
# create a custom feature engineering function 
def squ(x):
    return np.hstack((x,x**2))
# create FunctionTransformer to  apply  the  custom  function
custom_transformer  = FunctionTransformer(squ) 
# apply  the  transformer to  the  input data  
x_transformed  =  custom_transformer.transform(x)
# view the  transform  data  
print(x_transformed)  #   show  normal  array  and  its  squared  array
```

```python
# thrid example  

# craete  data  set 
s = np.array([[1,2],[3,4]])

# define a  custome  scaling  function
def  my_scaling(s):
    return s / np.max(s)
    
#create  a function  transformer  to  apply  the  custom  function  
customtransformer =  FunctionTransformer(my_scaling)

# apply the  transformer  to  the  input  data  
transformed_tf = customtransformer.transform(x)

# VIEW THE  TRANSFORMED  DATA  
print(transformed_tf)
```

```python
# Q.4  
# create data set  with  missing  value  
r = np.array([[1,2],[3,np.nan]])

# define  custom  cleaning  function  
def  my_cleaning(r):
    r[np.isnan(r)] = 0 
    return r
# create a function  TransFormer to  apply  the  custom function  
custom1 = FunctionTransformer(my_cleaning) 

# apply the  transformer  to the  inpute data 
x_transformed_2 = custom1.transform(r)
# view the  transformed  data 
print(x_transformed_2)
```
